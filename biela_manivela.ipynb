```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
O = np.array([0.0, 0.0])  # centro engrenagem

r = 1.0           # raio manivela
L = 4.0           # biela
y_slider = 0.3    # altura guia

n_frames = 240
```

```python
def mecanismo(theta):
    A = O + r * np.array([
        np.cos(theta),
        np.sin(theta)
    ])

    dy = y_slider - A[1]

    if abs(dy) > L:
        return None

    dx = np.sqrt(L**2 - dy**2)

    B = np.array([
        A[0] + dx,
        y_slider
    ])

    return A, B
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.set_aspect("equal")
ax.set_xlim(-2, 6)
ax.set_ylim(-3, 3)
ax.grid(True)

engrenagem = plt.Circle(O, r, fill=False, linewidth=2)
ax.add_patch(engrenagem)

ax.plot([-2, 6], [y_slider, y_slider], linewidth=3)   # guia

linha_manivela, = ax.plot([], [], linewidth=4)
linha_biela, = ax.plot([], [], linewidth=4)

ponto_O, = ax.plot([], [], "o", markersize=8)
ponto_A, = ax.plot([], [], "o", markersize=8)
ponto_B, = ax.plot([], [], "o", markersize=8)

cursor_w = 0.6
cursor_h = 0.35
cursor = plt.Rectangle((0, 0), cursor_w, cursor_h, fill=False, linewidth=2)
ax.add_patch(cursor)

def init():
    linha_manivela.set_data([], [])
    linha_biela.set_data([], [])
    ponto_O.set_data([], [])
    ponto_A.set_data([], [])
    ponto_B.set_data([], [])
    return linha_manivela, linha_biela, ponto_O, ponto_A, ponto_B, cursor

def update(frame):
    theta = 2 * np.pi * frame / n_frames

    resultado = mecanismo(theta)

    if resultado is None:
        return linha_manivela, linha_biela, ponto_O, ponto_A, ponto_B, cursor

    A, B = resultado

    linha_manivela.set_data([O[0], A[0]], [O[1], A[1]])
    linha_biela.set_data([A[0], B[0]], [A[1], B[1]])

    ponto_O.set_data([O[0]], [O[1]])
    ponto_A.set_data([A[0]], [A[1]])
    ponto_B.set_data([B[0]], [B[1]])

    cursor.set_xy((
        B[0] - cursor_w / 2,
        B[1] - cursor_h / 2
    ))

    return linha_manivela, linha_biela, ponto_O, ponto_A, ponto_B, cursor

anim = FuncAnimation(
    fig,
    update,
    frames=n_frames,
    init_func=init,
    interval=30,
    blit=True
)

plt.close(fig)

HTML(anim.to_jshtml())
```
